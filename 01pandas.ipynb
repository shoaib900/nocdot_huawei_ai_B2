```python
import numpy as np
import pandas as pd
```

```python
a = np.array([1, 2, 3])
b = np.array(9)
c = np.array([[1, 2, 3, 5, 6]])

print(a, a.ndim) # vector 1 ndim
print(b, b.ndim) # scalar 0 ndim
print(c, c.ndim) # matrix 2 ndim
```

```python
a = np.zeros((3, 4))

print(a, a.ndim) # matrix 2 ndim
```

```python
b = np.ones((3, 4))

print(b, b.ndim) # matrix 2 ndim
```

```python
c = np.eye(3)

print(c, c.ndim) # matrix 2 ndim
```

```python
d = np.random.rand(3, 4)

print(d, d.ndim) # matrix 2 ndim
```

```python
e = np.arange(10)

print(e, e.ndim) # vector 1 ndim
```

```python
f = np.linspace(0, 1, 7)

print(f, f.ndim) # vector 1 ndim
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

g = np.sum([a, b], axis=0)

print(g, g.ndim) # vector 1 ndim
```
