```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
```

```python
# house prediction dataset
X = np.array([[1], [2], [3], [4], [5]]) # 2 dimensional array  / area
y = np.array([2, 4, 6, 8, 10]) # 1 dimensional array / price

model = LinearRegression()
model.fit(X, y) # independent variable and dependent variable

prediction = model.predict(np.array([[6]])) # predict the price of house with area 6

print("Predicted price of house with area 6:", prediction[0])
```
